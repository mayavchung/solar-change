# Load the feedbacks data and save versions with only the global mean fields

```python
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
import netCDF4
```

```python
def get_gm(ds):
    # get the global mean variables only
    gm_vars = [v for v in ds.data_vars if "_gm" in v]
    ds_new = ds[gm_vars]
    return ds_new
    
```

```python
exps = [f"{sign}{pct}%solar" 
        for pct in [1, 2, 4, 6] 
        for sign in ["-", "+"]]
```

```python
for ii in range(np.size(exps)):
    exp = exps[ii]
    print(exp)
    
    # load data
    print('loading...')
    FLOR_files = f'./rk_decomp_cloud/rk.decomp.FLOR.{exp}.50yrref.*.nc'
    FLOR_rk = xr.open_mfdataset(FLOR_files, combine="by_coords", parallel=False)
    CM21_files = f'./rk_decomp_cloud/rk.decomp.CM2.1p1.{exp}.50yrref.*.nc'
    CM21_rk = xr.open_mfdataset(CM21_files, combine="by_coords", parallel=False)

    # get global means only
    FLOR_rk_gm = get_gm(FLOR_rk)
    CM21_rk_gm = get_gm(CM21_rk)

    # save for Zenodo
    print('saving...')
    FLOR_rk_gm.to_netcdf(f'./gm_dR_zenodo/rk.decomp.FLOR.{exp}.50yrref.gm.nc')
    CM21_rk_gm.to_netcdf(f'./gm_dR_zenodo/rk.decomp.CM2.1p1.{exp}.50yrref.gm.nc')
```
